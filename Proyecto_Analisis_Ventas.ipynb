```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

print("Bibliotecas cargadas correctamente")
```

```python
np.random.seed(42)
n = 1000
```

```python
datos = pd.DataFrame({
  "Fecha": pd.date_range(
      start="2024-01-01",
      periods=n,
      freq="D"

     ),
  "Producto": np.random.choice(
      ["Laptop", "Mouse", "Teclado", "Monitor"],
      n



  ),
    "Cantidad": np.random.randint(1, 11, n),
    "Precio_Unitario": np.random.randint(100, 15000, n

  )
})

datos["Total_Ventas"] = (
    datos["Cantidad"] * datos["Precio_Unitario"]

)

datos.to_csv("ventas_empresa.csv", index=False)
datos.head()
```

```python
ventas_producto =(
    datos.groupby("Producto")["Total_Ventas"]
    .sum()
    .sort_values(ascending=False)
)
print(ventas_producto)

ventas_producto.plot(
   kind="bar",
   figsize=(9,5),
   title="Ventas totales por producto"
)
plt.xlabel("producto")
plt.ylabel("Ventas en MXN")
plt.xticks(rotation=0)
plt.show()




```
